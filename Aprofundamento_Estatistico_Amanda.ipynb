# Identificação do Problema

Durante a análise exploratória, foi observada uma forte correlação positiva entre a temperatura do ar e a temperatura do processo.

O objetivo desta análise é verificar se essa relação é estatisticamente significativa ou se pode ter ocorrido ao acaso.

## Teste de Normalidade

```python
from scipy.stats import shapiro

stat_air, p_air = shapiro(df['Air temperature [K]'])
stat_process, p_process = shapiro(df['Process temperature [K]'])

print("Air temperature p-value:", p_air)
print("Process temperature p-value:", p_process)
```

Air temperature p-value: 3.16e-33
Process temperature p-value: 3.16e-22
Ambos são p < 0,05, ou seja, os dados não seguem uma distribuição normal. Por isso vamos usar testes não paramétricos, aqui no caso, o Spearmen

## Formulação de Hipóteses

H0 (Hipótese Nula): Não existe correlação entre a temperatura do ar e a temperatura do processo.

H1 (Hipótese Alternativa): Existe correlação entre a temperatura do ar e a temperatura do processo.

##Execução do Teste Estatístico

```python
from scipy.stats import spearmanr

corr, p_value = spearmanr(df['Air temperature [K]'], df['Process temperature [K]'])

print("Correlação:", corr)
print("p-value:", p_value)
```

Foi utilizado o teste de correlação de Spearman, devido à ausência de normalidade nos dados.

O resultado apresentou uma correlação de 0.86, indicando uma forte relação positiva entre a temperatura do ar e a temperatura do processo.

O tamanho do efeito foi avaliado com base no coeficiente de correlação de Spearman, que apresentou valor de 0.86.

Esse valor indica um efeito forte, demonstrando que a relação entre as variáveis analisadas é significativa não apenas estatisticamente, mas também na prática.

O p-valor encontrado foi inferior a 0,05 (p ≈ 0.0), permitindo rejeitar a hipótese nula.


## Conclusão Científica

Com base nos resultados obtidos, conclui-se que a correlação observada entre a temperatura do ar e a temperatura do processo é estatisticamente significativa, não sendo fruto do acaso. Podendo assim, descartar a hipótese nula.

Além disso, o valor da correlação indica um efeito forte, demonstrando relevância prática para o problema analisado.

No contexto do projeto, essa relação sugere que a temperatura do ar é um fator importante para prever o comportamento da temperatura do processo, sendo boas features, podendo ser utilizada como variável relevante em modelos futuros de Machine Learning. é preciso ter atenção porque as vezes a correlação muito alta de duas variáveis pode acarretar na duplicidade de dados, porém em contrapartida, aumentam a capacidade preditiva do modelo.

Foi utilizado o teste de correlação de Spearman, pois o objetivo da análise foi verificar a relação entre duas variáveis contínuas.

Os testes como Teste-t, ANOVA e Mann-Whitney não foram utilizados, pois são aplicados em comparações entre grupos, o que não corresponde ao objetivo desta análise.
